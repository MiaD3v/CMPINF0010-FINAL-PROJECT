# Pittsburgh 311 Data

## What Are We Looking For?

We are looking for the neighborhoods with the largest amounts of illegal dumping complaints!

## Getting Started

Let's create our dataframe here!

```python
import pandas as pd
import numpy as np

dtype = {
    "street": np.str_,
    "street_id": np.str_,
    "neighborhood": np.str_,
}

df = pd.read_csv("data.csv", dtype=dtype, parse_dates=True)

df
```

## We Want Pittsburgh ***Only***!

Let's filter the "city" column by "city == Pittsburgh"

```python
df = df[df["city"] == "Pittsburgh"]
df
```

## Illegal Dumping Complaints!

Now, because we are looking for the neighborhood with the largest amount of illegal dumping, let's filter "subject_code" == "ILLEDUMP"!

```python
df = df[df["subject_code"] == "ILLEDUMP"]

df
```

## We're almost there!

Now we must see how many times we see each neighborhood; the more the better!

```python
val_count = df["neighborhood"].value_counts()
val_count
```
