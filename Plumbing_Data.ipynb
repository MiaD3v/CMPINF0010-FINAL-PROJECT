##Overview 
This data set looks at plumbing inspections failed or passed across allegheny county, the data has been cleaned to only include residential buildings in the greater Pittsburgh area. 

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("CMPINF0010_Lab_Data_Plumbing.csv")
df=df[(df["city"]!="PITTSBURGH")&(df["building_type"]!="RESIDENTIAL")]
df.head()

n_count=(
    df[df["final_insp_yn"]=="N"]
    .groupby("municipal")
    .size()
    .sort_values(ascending=False)
)

n_count.head(10).plot(kind="bar",figsize=(10,5))
plt.ylabel("Failed inspections")
plt.title("Failed inspections by neighborhood")
plt.tight_layout()
plt.show()
```

## Conclusion
Moon Township has the greatest number of failed inspections across all neighborhoods in the Pittsburgh area. By this metric alone Moon Township would be the best neighborhood to live in for a computer science student. It should be noted that this is without doing a proportion test.
